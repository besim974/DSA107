this was gemini's answer :

Missing sex determinations in penguin field datasets (such as the commonly used Palmer Station / Palmer Penguins dataset collected by Dr. Kristen Gorman) typically stem from four biological, logistical, and technical reasons:

Lack of Sexual Dimorphism (Visual Similarity):

Most penguin species (Adélie, Chinstrap, Gentoo) are monomorphic—males and females look nearly identical externally. They do not have external genitalia, and while males are slightly larger on average, individual size ranges overlap significantly. Sex cannot be reliably determined by visual inspection alone.

Inconclusive or Failed Molecular/Blood Testing:

Definitive sex determination usually requires molecular sexing via PCR amplification of CHD genes from a blood sample.

Samples can degrade in harsh Antarctic conditions, yield insufficient DNA, or produce ambiguous PCR bands that researchers exclude to avoid recording false data.

Logistical and Animal Welfare Constraints:

Field researchers prioritize animal welfare. If an individual penguin becomes excessively stressed, if weather conditions deteriorate suddenly, or if capturing a blood sample presents a risk of injury to the bird or handler, researchers will take only basic external measurements (bill length, flipper length, body mass) and skip invasive blood sampling.

Juveniles or Molting Individuals:

Younger or immature birds have not fully developed adult physical proportions, and taking invasive samples from chicks or molting individuals is often restricted or avoided in observational protocols.
